```python
import pandas as pd
import numpy as np
from sklearn.ensemble import IsolationForest
```

```python
data={
    'login-attempt':[2,3,5,7,2,9],
    'access-time':[11,13,15,2,15,3],
    'files-accessed':[4,6,8,100,5,150]
}
df=pd.DataFrame(data)
```

```python
model=IsolationForest(n_estimators=100, contamination=0.2, random_state=40)
model.fit(df)

#redict: -1 means anomaly else no anomaly
df['threat detected']=model.predict(df)
print(df)

anomalies=df[df['threat detected']==-1]
print("\n possbible threat detected:")
print(anomalies)
```
